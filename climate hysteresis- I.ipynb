```python
def global_annual_mean(ds, var):
    data = ds[var]

    # step 1: yearly mean (reduces time dimension)
    data = data.groupby('time.year').mean()

    # step 2: weights
    weights = np.cos(np.deg2rad(data['lat']))
    weights = weights / weights.mean()

    # step 3: spatial mean
    return data.weighted(weights).mean(dim=['lat', 'lon'])
```

```python
def global_mean(ds, var):
    data = ds[var]
    weights = np.cos(np.deg2rad(data['lat']))
    weights = weights / weights.mean()
    return data.weighted(weights).mean(dim=['lat', 'lon'])
```

```python
ts_combined = xr.concat([ts_ru, ts_rd[1:]], dim='year')
pr_combined = xr.concat([pr_ru, pr_rd[1:]], dim='year')

total_len = len(ts_combined)
years = np.arange(total_len)
```

```python
C0 = 300
Cmax = 4 * C0

t_peak = 140
t_return = 280

# --- ramp-up ---
co2_up = C0 * np.exp(np.linspace(0, np.log(4), t_peak))

# --- ramp-down ---
co2_down = co2_up[::-1]

# --- flat after 280 ---
remaining_len = total_len - (len(co2_up) + len(co2_down))
co2_flat = np.full(max(0, remaining_len), C0)

# --- combine ---
co2 = np.concatenate([co2_up, co2_down, co2_flat])

# match exact length
co2 = co2[:total_len]

peak_idx = t_peak - 1
```
